# Zwischenfazit: 
# Dänemark führt das Ranking mit 42,03 % an, es folgen Finnland und Schweden. 
# Rumänien ist mit 5,21 % Schlusslicht.

```python
# df für spannweite über jahre erstellen
df_spannweite = df_eu.groupby("jahr")["ki_nutzung_prozent"].agg(["min", "max"])
df_spannweite["range"] = df_spannweite["max"] - df_spannweite["min"]
df_spannweite.columns = ["min(%)", "max(%)", "SW(Pp.)"]
print(df_spannweite)
```

```python
# df für mittelwert und standardabweichung pro jahre erstellen 
df_durchschnitt = df_eu.groupby("jahr")["ki_nutzung_prozent"].mean()
df_durchschnitt = df_durchschnitt.to_frame("MW(%)")
df_durchschnitt["SD(Pp.)"] = df_eu.groupby("jahr")["ki_nutzung_prozent"].std(ddof=0)
df_durchschnitt["VK(%)"] = df_durchschnitt["SD(Pp.)"] / df_durchschnitt["MW(%)"] * 100
print(df_durchschnitt)
```

```python
# df für Interquartilsabstände pro Jahr erstellen
df_iqr = df_eu.groupby("jahr")["ki_nutzung_prozent"].quantile(0.25)
df_iqr = df_iqr.to_frame("Q1(%)")
df_iqr["Q3(%)"] = df_eu.groupby("jahr")["ki_nutzung_prozent"].quantile(0.75)
df_iqr["IQR(Pp.)"] = df_iqr["Q3(%)"] - df_iqr["Q1(%)"]
print(df_iqr)
```

# Zwischenfazit: 
# Spannweite, Standardabweichung und Interquartilsabstand nehmen zwischen 2021 und 2025 deutlich zu. 
# Zugleich sinkt der Variationskoeffizient. 
# Ergo: Die absolute Kluft zwischen den Ländern wächst, während die relative Streuung sinkt.

```python
# Zur Kontrolle der Werte df ohne Frankreich, Schweden, Niederlande wegen Eintrag Zeitreihenbruch 
df_eu_sens = df_eu[~df_eu["land"].isin(["Frankreich", "Schweden", "Niederlande"])]
```

```python
# df für spannweite_sens über jahre erstellen
df_spannweite_sens = df_eu_sens.groupby("jahr")["ki_nutzung_prozent"].agg(["min", "max"])
df_spannweite_sens["range"] = df_spannweite_sens["max"] - df_spannweite_sens["min"]
df_spannweite_sens.columns = ["min(%)", "max(%)", "SW(Pp.)"]
print(df_spannweite_sens)
```

```python
# df sens für mittelwert und standardabweichung pro jahre erstellen 
df_durchschnitt_sens = df_eu_sens.groupby("jahr")["ki_nutzung_prozent"].mean()
df_durchschnitt_sens = df_durchschnitt_sens.to_frame("MW(%)")
df_durchschnitt_sens["SD(Pp.)"] = df_eu_sens.groupby("jahr")["ki_nutzung_prozent"].std(ddof=0)
df_durchschnitt_sens["VK(%)"] = df_durchschnitt_sens["SD(Pp.)"] / df_durchschnitt_sens["MW(%)"] * 100
print(df_durchschnitt_sens)
```

```python
# df sens für Interquartilsabstände pro Jahr erstellen
df_iqr_sens = df_eu_sens.groupby("jahr")["ki_nutzung_prozent"].quantile(0.25)
df_iqr_sens = df_iqr_sens.to_frame("Q1(%)")
df_iqr_sens["Q3(%)"] = df_eu_sens.groupby("jahr")["ki_nutzung_prozent"].quantile(0.75)
df_iqr_sens["IQR(Pp.)"] = df_iqr_sens["Q3(%)"] - df_iqr_sens["Q1(%)"]
print(df_iqr_sens)
```

### Zwischenfazit im Vergleich mit eu_sens: 
# Spannweite: bleibt vollständig unverändert.
# Standardabweichung: steigt weiterhin stark.
# Variationskoeffizient: sinkt weiterhin deutlich im Vgl. zu 2021, steigt allerdings zwischen 2024 und 2025 etwas.
# Interquartilsabstand: steigt weiterhin, allerdings weniger stark als in der Analyse mit allen Ländern.
# Es zeichnet sich ähnliches Bild: absolute Streuung nimmt zu, während  relative Streuung im Vgl. zu 2021 abnimmt.

```python
# df_eu_sens für Zeitvergleich in Pivot-tabelle umwandeln
df_eu_sens_wide = df_eu_sens.pivot(index="land", columns="jahr", values="ki_nutzung_prozent")
print(df_eu_sens_wide)
```

```python
# Berechne Zuwachs der Nutzung 2021 zu 2025
df_eu_sens_wide["abs. Zuwachs(Pp.)"] = df_eu_sens_wide[2025] - df_eu_sens_wide[2021]
df_eu_sens_wide["rela. Zuwachs(%)"] = (df_eu_sens_wide[2025] - df_eu_sens_wide[2021]) / df_eu_sens_wide[2021] * 100
print(df_eu_sens_wide)
```

```python
# Top-5 absoluter Zuwachs
print(df_eu_sens_wide.nlargest(5, "abs. Zuwachs(Pp.)"))
```

```python
# Top-5 relativert Zuwachs
print(df_eu_sens_wide.nlargest(5, "rela. Zuwachs(%)"))
```

```python
print(df_eu_sens_wide[2025].nsmallest(5))
```

# Zwischenfazit: 
# Zwei Länder sind besonders interessant:
# - Estland: Trotz niedriger Ausgangsnutzung ist es absolut und relativ unter den Top-5-Wachstums-Ländern.
# - Rumänien: Trotz relativem Zuwachs von ca. 278 % ist das Land mit KI-Nutzung von 5,21 % Schlusslicht. 

```python
# Noch prüfen, warum zwischen 2021 und 2023 alle Streuungsmaße sinken
df_eu_sens_wide["Zuwachs_21_23(Pp.)"] = df_eu_sens_wide[2023] - df_eu_sens_wide[2021]
print(df_eu_sens_wide)
```

```python
# alle Länder mit gesunkenem Nutzungsantail zw. 2021 und 2023
print(df_eu_sens_wide[df_eu_sens_wide["Zuwachs_21_23(Pp.)"] < 0])
```

```python
# Top-5 abs. Zuwachs zw. 2021 und 2023
print(df_eu_sens_wide.nlargest(5, "Zuwachs_21_23(Pp.)"))
```

# Zwischenfazit:
# Die sinkende Streuung zwischen 2021 und 2023 hängt mit sinkender Nutzung in den Top-Ländern, insb. Dänemark, zusammen.
# Und mit Zuwächsen in Ländern mit niedriger Ausgangsnutzung. 
# Was die jeweiligen Gründe in den Ländern für die Nutzungsänderung sind, ist noch offen. 

```python
# Ranking hinzufügen für Länder in Jahren 2021 und 2025 für Vergleiche und sortieren
df_eu_sens_wide["Rang_2021"] = df_eu_sens_wide[2021].rank(ascending=False).astype(int)
df_eu_sens_wide["Rang_2025"] = df_eu_sens_wide[2025].rank(ascending=False).astype(int)
df_eu_sens_wide["Rangveraenderung"] = df_eu_sens_wide["Rang_2021"] - df_eu_sens_wide["Rang_2025"]
df_eu_sens_wide = df_eu_sens_wide.sort_values(by = "Rangveraenderung", ascending=False)

print(df_eu_sens_wide)
```

# Zwischenfazit:
# Estland ist mit +14 der größte Aufsteiger unter den 24 Ländern, es folgt Litauen (+6). 
# Kroatien und Portugal verlieren die meisten Plätze.
# Dänemark, Finnland und Rumänien behalten ihre Plätze an der Spitze bzw. am Ende der Rangliste.

```python
# für Grafiken matplotlib importieren
import matplotlib.pyplot as plt
```

```python
# Daten sortieren
df_plot = df_eu_sens_wide.sort_values(by=2025)

# Grafik erstellen
plt.figure(figsize=(10, 9))

# Breite, helle Balken für 2025
plt.barh(
    df_plot.index,
    df_plot[2025],
    height=0.8,
    color="lightblue",
    label="2025"
)

# Schmale, dunkle Balken für 2021
plt.barh(
    df_plot.index,
    df_plot[2021],
    height=0.4,
    color="steelblue",
    label="2021"
)

# Titel und Untertitel
plt.suptitle(
    "KI-Nutzung nimmt in allen betrachteten Ländern zu",
    fontsize=18,
    fontweight="bold"
)

plt.title(
    "Vergleich von 24 EU-Ländern zwischen 2021 und 2025",
    fontsize=12
)

# x-Achse beschriften
plt.xlabel("Anteil der Unternehmen mit KI-Nutzung (%)")

# Legende
plt.legend(loc="lower right")

# Quellenangabe und methodischer Hinweis
plt.figtext(
    0.01, 0.01,
    "Quelle: Eurostat (isoc_eb_ai), eigene Auswertung\n"
    "Ohne Frankreich, Schweden und die Niederlande aufgrund von Zeitreihenbrüchen.",
    fontsize=9
)

#Grafik sichern 
plt.savefig(
    "output/ki_ranking_2025.png",
    dpi=200,
    bbox_inches="tight"
)
# Grafik anzeigen
plt.show()
```

```python
# Balkendiagramm erstellen, dass Rangveränderung von 2021 bis 2025 zeigt:
# Farben festlegen
farben = []

for wert in df_plot["Rangveraenderung"]:
    if wert > 0:
        farben.append("#79B473")
    elif wert < 0:
        farben.append("#D97B7B")
    else:
        farben.append("#D3D3D3")

# Grafik erstellen
plt.figure(figsize=(10, 9))

# Balkendiagramm
plt.barh(
    df_plot.index,
    df_plot["Rangveraenderung"],
    color=farben
)

# Nulllinie einzeichnen
plt.axvline(x=0, color="black", linewidth=0.8)

# Länder ohne Rangveränderung filtern
df_null = df_plot[df_plot["Rangveraenderung"] == 0]

# Diese Länder als Punkt einzeichnen
plt.scatter(
    df_null["Rangveraenderung"],
    df_null.index,
    color="grey",
    zorder=3,
    label="= keine Rangveränderung"
)

# Titel und Untertitel 
plt.suptitle(
    "Estland steigt im KI-Ranking am stärksten auf",
    fontsize=18,
    fontweight="bold"
)
plt.title(
    "Veränderung im EU-Ranking des KI-Nutzungsanteils in Unternehmen, 2021–2025",
    fontsize=12
)

# x-Achse beschriften
plt.xlabel("Gewonnene bzw. verlorene Rangplätze")
plt.xticks(range(-14, 15, 2))
plt.xlim(-15, 15)

# Quellenangabe
plt.figtext(
    0.01, 0.01,
    "Quelle: Eurostat (isoc_eb_ai), eigene Auswertung\n"
    "Ohne Frankreich, Schweden und die Niederlande aufgrund von Zeitreihenbrüchen.",
    fontsize=9
)

# Legende für graue Punkte - ohne Rangänderung - hinzufügen
plt.legend(loc="lower right", frameon=False)

#Grafik sichern
plt.savefig(
    "output/rangveraenderungen.png",
    dpi=200,
    bbox_inches="tight"
)
# Grafik anzeigen
plt.show()
```
